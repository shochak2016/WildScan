```python
import os
import pandas as pd

base_path = "../Animal-Sound-Dataset"

rows = []

for label in os.listdir(base_path):
    folder = os.path.join(base_path, label)
    
    if not os.path.isdir(folder):
        continue
    
    for file in os.listdir(folder):
        if file.endswith(".wav"):
            rows.append({
                "file_path": os.path.join(folder, file),
                "label": label
            })

df = pd.DataFrame(rows)

print(df.head())
```

```python
def load_audio(path):
    try:
        with warnings.catch_warnings():
            warnings.simplefilter("ignore")  # hides ugly audio warnings
            y, sr = librosa.load(path, sr=16000)
        return y, sr
    except Exception:
        return None, None


df["audio"] = df["file_path"].apply(lambda p: load_audio(p)[0])
df["sr"] = df["file_path"].apply(lambda p: load_audio(p)[1])
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
df
```
