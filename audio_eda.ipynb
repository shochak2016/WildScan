```python
import os
import pandas as pd

base_path = "../Animal-Sound-Dataset"

rows = []

for label in os.listdir(base_path):
    folder = os.path.join(base_path, label)
    
    if not os.path.isdir(folder):
        continue
    
    for file in os.listdir(folder):
        if file.endswith(".wav"):
            rows.append({
                "file_path": os.path.join(folder, file),
                "label": label
            })

df = pd.DataFrame(rows)

print(df.head())

#Put all paths to each individual path into pandas 
```

```python
base_path = "../Animal-Sound-Dataset" #This works under the assumption that the dataset is in the same folder as the repo, change accordingly

spectrograms = []

for root, _, files in os.walk(base_path):
    for file in files:
        if file.endswith(".wav"):
            path = os.path.join(root, file)

            # load audio
            y, sr = librosa.load(path, sr=16000)

            # mel spectrogram
            S = librosa.feature.melspectrogram(y=y, sr=sr)
            S_db = librosa.power_to_db(S, ref=np.max)

            spectrograms.append(S_db)

print("Loaded spectrograms:", len(spectrograms))

```
